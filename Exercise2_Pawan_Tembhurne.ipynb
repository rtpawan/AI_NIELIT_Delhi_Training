```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Load two sample images
image1 = cv2.imread("img1.jpg")
image2 = cv2.imread("img2.jpg")

# Check whether images are loaded successfully
if image1 is None:
    raise FileNotFoundError("image1.jpg not found!")

if image2 is None:
    raise FileNotFoundError("image2.jpg not found!")
```

```python
# Step 2: Resize both images to same dimensions
width = 500
height = 400

image1 = cv2.resize(image1, (width, height))
image2 = cv2.resize(image2, (width, height))
```

```python
# Step 3: Perform weighted image blending
# alpha = weight of first image
# beta = weight of second image
# gamma = brightness adjustment

blended_image = cv2.addWeighted(
    image1, 0.5,
    image2, 0.5,
    0
)
```

```python
# Step 4: Save the output image
cv2.imwrite("blended_output.jpg", blended_image)

print("Image blending completed successfully!")
print("Output saved as blended_output.jpg")
```

```python
# Step 5: Convert BGR to RGB for displaying with Matplotlib
image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended_image, cv2.COLOR_BGR2RGB)
```

```python
# Step 6: Display original and blended images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(image2_rgb)
plt.title("Original Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
